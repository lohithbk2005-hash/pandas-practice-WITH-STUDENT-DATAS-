```python
import pandas as pd
import numpy as np
```

```python
people={
    'name':['Raj','Dev','Lohi',None,np.nan,'Na'],
    'age':[20,30,21,np.nan,None,'Missing'],
    'last':['a','c','b',None,np.nan,'Missing']
}
```

```python
df=pd.DataFrame(people)
```

```python
df
```

```python
df.dropna(axis='index',how='all',subset=['name','age'])
```

```python
df.isna()
```

```python
df.replace('Na',np.nan,inplace=True)
df.replace('Missing',np.nan,inplace=True)
```

```python
df.dropna(axis='index',how='all',subset=['last'])
```

```python
df['age'].dtype
```

```python
df['age']=df['age'].astype('float')
```

```python
df['age'].mean()
```
